Run only after all selected discovery and Predictor jobs complete. This evaluates fixed final checkpoints and includes output-specific ID ridge/logistic and zero/shuffle diagnostics. Stored results were generated by execution on 30 September 2026 and checked against the submitted manuscript. The supplementary C39 variant is trained before evaluation below.

```python
from pathlib import Path
import json, subprocess, sys, shutil
ROOT = Path.cwd().resolve()
if ROOT.name == 'notebooks': ROOT = ROOT.parent
assert (ROOT/'code/train.py').is_file(), 'Open from repository root or notebooks/'
print('Repository:', ROOT.name)
def run_command(args, cwd=None, check=True):
    """Stream subprocess output into the notebook, preserving failures."""
    proc = subprocess.Popen(args, cwd=cwd, stdout=subprocess.PIPE,
                            stderr=subprocess.STDOUT, text=True, bufsize=1)
    for line in proc.stdout:
        print(line.replace(str(ROOT), '.'), end='', flush=True)
    code = proc.wait()
    if check and code:
        raise RuntimeError(f'Command failed with exit code {code}')

```

```python
RUN=ROOT/'work/manuscript_run'
assert (RUN/'train_complete.json').exists(), 'Complete notebook 02 first'
```

```python
# Include the published supplementary C39 readout variant.
# Completed direct runs are retained; only missing runs are trained.
cfg = json.loads((RUN/'plan.json').read_text())
cfg['conditions'] = ['direct', 'C39']
(RUN/'plan.json').write_text(json.dumps(cfg, indent=2))
run_command([sys.executable, str(RUN/'run_all.py'), '--execute', '--jobs', '2'], cwd=RUN)
```

```python
run_command([sys.executable,str(RUN/'run_all.py'),'--evaluate'],cwd=RUN,check=True)
```
