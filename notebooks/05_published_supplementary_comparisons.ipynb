# Published supplementary beach comparisons
Reproduce the Joint column and the two other discovery-output rows already present in submitted Supplement S11. This does not add a new comparison to the manuscript. Uses only fit/validation arrays, never test. The same code and original per-experiment plans are retained.


```python
from pathlib import Path
import json, subprocess, sys, shutil
ROOT = Path.cwd().resolve()
if ROOT.name == 'notebooks': ROOT = ROOT.parent
assert (ROOT/'code/train.py').is_file(), 'Open from repository root or notebooks/'
print('Repository:', ROOT.name)
def run_command(args, cwd=None, check=True):
    """Stream subprocess output into the notebook, preserving failures."""
    proc = subprocess.Popen(args, cwd=cwd, stdout=subprocess.PIPE,
                            stderr=subprocess.STDOUT, text=True, bufsize=1)
    for line in proc.stdout:
        print(line.replace(str(ROOT), '.'), end='', flush=True)
    code = proc.wait()
    if check and code:
        raise RuntimeError(f'Command failed with exit code {code}')

```

```python
BASE=ROOT/'work/beach_supplement'
BASE.mkdir(parents=True,exist_ok=True)
JOBS=[('attack_success','joint',['blocks_per_set','digs_per_set']),('attack_efficiency','mse',['digs_per_set']),('point_share','mse',['digs_per_set'])]
for variant,condition,targets in JOBS:
    folder=BASE/variant
    if folder.exists():
        raise FileExistsError('Use a fresh directory for staging, or rerun only the training cell to resume.')
    shutil.copytree(ROOT/'extensions/beach_volleyball',folder)
    shutil.copy2(ROOT/f'extensions/beach_volleyball/plans/{variant}.json',folder/'plan.json')
    for kind in ['source','reuse']:
        (folder/'data'/kind).mkdir(parents=True)
        for name in ['fit.npz','val.npz','meta.json']:
            shutil.copy2(ROOT/f'work/raw_build/beach/{variant}/data/{kind}/{name}',folder/'data'/kind/name)
```

```python
for variant,condition,targets in JOBS:
    folder=BASE/variant
    for seed in [17,43,101]:
        run_command([sys.executable,str(folder/'run_job.py'),'finder',condition,str(seed)],cwd=folder)
        for target in targets:
            run_command([sys.executable,str(folder/'run_job.py'),'predictor',condition,str(seed),'--target',target],cwd=folder)
```

```python
results={}
for variant,condition,targets in JOBS:
    results[variant]={}
    for stage in ['finder']+['predictor_'+target for target in targets]:
        values=[]
        for seed in [17,43,101]:
            record=json.loads((BASE/f'{variant}/runs/seed_{seed}/{condition}/{stage}/best.json').read_text())
            values.append(record['validation_mean_AB'] if stage=='finder' else record['validation_loss'])
        results[variant][stage]={'seed_losses':values,'mean':sum(values)/len(values)}
print(json.dumps(results,indent=2))
(BASE/'summary.json').write_text(json.dumps(results,indent=2));
```

```python
# Reproduce the arithmetic of the originally published supplementary table.
# best.json uses float64 selection metrics; the historical table aggregates
# squared errors from saved float32 predictions. Do not conflate the two.
from pathlib import Path
import numpy as np
root=Path.cwd()
if root.name=='notebooks':root=root.parent
expected={('attack_success','finder'):'0.00662683',('attack_success','predictor_blocks_per_set'):'0.89568895',('attack_success','predictor_digs_per_set'):'4.80233049',('attack_efficiency','predictor_digs_per_set'):'4.87480879',('point_share','predictor_digs_per_set'):'4.84798384'}
for (variant,stage),reported in expected.items():
    condition='joint' if variant=='attack_success' else 'mse'
    losses=[]
    for seed in [17,43,101]:
        z=np.load(root/f'work/beach_supplement/{variant}/runs/seed_{seed}/{condition}/{stage}/validation_predictions.npz')
        pred,y=z['prediction'],z['y']
        losses.append(((pred-y[:,None])**2).mean(1) if pred.ndim==2 else (pred-y)**2)
    observed=float(np.mean(losses))
    print(variant,condition,stage,'manuscript:',reported,'rerun:',f'{observed:.8f}','MATCH' if f'{observed:.8f}'==reported else 'DIFFERENCE: inspect platform/precision and source versions')
```
